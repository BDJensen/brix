```python
my_storageaccount = 'openuniverselake'
my_container = 'ais2021'
my_mountpoint = '/mnt/bd/openuniverselake'

my_sastoken = dbutils.widgets.get("my_sastoken")
```

```python
# dbutils.fs.unmount(my_mountpoint)
dbutils.fs.mounts()
```

```python
source = f"wasbs://{my_container}@{my_storageaccount}.blob.core.windows.net"
configs = {f"fs.azure.sas.{my_container}.{my_storageaccount}.blob.core.windows.net": my_sastoken}
print(f"{source = }")
print(f"{configs = }")
```

```python
dbutils.fs.mount(source = source, mount_point = my_mountpoint, extra_configs = configs)
dbutils.fs.ls(f"dbfs:{my_mountpoint}")
```
