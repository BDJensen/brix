```sql
CREATE WIDGET TEXT schema_name DEFAULT "vessel";
CREATE WIDGET TEXT delta_path DEFAULT "/mnt/bd/openuniverselake/aisraw/aisdata/delta";
```

```sql
SELECT CASE WHEN latitude<-90.0 THEN 1 ELSE 0 END lat_too_low
     , CASE WHEN latitude>90.0  THEN 1 ELSE 0 END lat_too_high
     , CASE WHEN longitude<-180.0 THEN 1 ELSE  0 END long_too_low
     , CASE WHEN longitude>180.0 THEN 1 ELSE  0 END long_too_high
     , count(*) cnt
FROM ${schema_name}.messages
GROUP BY CASE WHEN latitude<-90.0 THEN 1 ELSE   0 END
       , CASE WHEN latitude>90.0  THEN 1 ELSE   0 END
       , CASE WHEN longitude<-180.0 THEN 1 ELSE 0 END
       , CASE WHEN longitude>180.0 THEN 1 ELSE  0 END
;
```

```sql
SELECT COUNT(*) CNT, COUNT(DISTINCT *) uniq FROM ${schema_name}.messages;
/* 390.251.689	41.131.339 :-0  7min*/
```

```sql
SELECT yyyy, yyyy_mm, yyyy_mm_dd, min(rec_time) min_rec_time , max(rec_time) max_rec_time, COUNT(*) cnt, count(distinct *) uniq
FROM ${schema_name}.messages
GROUP BY  yyyy, yyyy_mm, yyyy_mm_dd
ORDER BY 1;
```
