```sql
CREATE OR REPLACE VIEW vessel.numbers
AS
SELECT t.nbr0 as nbr
FROM ( SELECT posexplode(sequence(1, 1000001)) AS (nbr0, nbr) ) t;


CREATE OR REPLACE VIEW vessel.letters
AS
SELECT char(n) AS letter, upper(char(n)) as big_letter
FROM (SELECT sequence(97, 122) AS letters) t
LATERAL VIEW explode(letters) l AS n;
```
