```python
# generated code by asking "write python code to list files from this web folder http://aisdata.ais.dk/?prefix=2025/"

import requests
from bs4 import BeautifulSoup

url = "http://aisdata.ais.dk/?prefix=2025/"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

files = []
for link in soup.find_all("a"):
    href = link.get("href")
    if href and href.startswith("2025"):
        files.append(href)

display(files)
```

```python
# create a list of dates for all days in january 2025, following this pattern yyyy-mm-dd, f. ex. 2025-01-01
from datetime import date, timedelta

start_date = date(2025, 1, 1)
end_date = date(2025, 1, 31)
dates_jan_2025 = [(start_date + timedelta(days=i)).strftime("%Y-%m-%d") for i in range((end_date - start_date).days + 1)]

print(dates_jan_2025)
#display([(date_str,) for date_str in dates_jan_2025])
```
