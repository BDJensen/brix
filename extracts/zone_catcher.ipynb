```python
import geopandas as gpd
import requests
import io
```

```python
# WFS URL returnerer GeoJSON

wfs_url = "https://havplan.dk/geoserver/havplan/ows"

params = {

    "service": "WFS",

    "version": "1.0.0",

    "request": "GetFeature",

    "typeName": "havplan:data",

    "outputFormat": "application/json"

}

 

# Load GeoDataFrame fra WFS

gdf = gpd.read_file(wfs_url, params=params)

 

# Konverter geometri til WKT

gdf["wkt"] = gdf["geometry"].apply(lambda geom: geom.wkt)

 

# Gem som CSV

gdf.to_csv("havplan_wfs_data.csv", index=False)
```
