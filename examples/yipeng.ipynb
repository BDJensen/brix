```sql
SELECT * 
FROM IDENTIFIER(concat(:schema_name, '.messages'))
WHERE name LIKE 'YI PENG 3'
AND yyyy_mm= '2024-11';
```

```sql
SELECT x.*
FROM (
      SELECT *
          , row_number() over (partition by mmsi, date_format(rec_time, 'yyyy-MM-dd HH:mm')  order by rec_time DESC) rnr
      FROM IDENTIFIER(concat(:schema_name, '.messages'))
      WHERE name LIKE 'YI PENG 3'
      AND yyyy_mm= '2024-11'
) x
WHERE x.rnr = 1
ORDER BY x.rec_time;
```
